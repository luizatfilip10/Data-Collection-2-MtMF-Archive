# 1. Introduction 

## 1.1 What is the Movies that Matter Festival?

The <a href="https://moviesthatmatter.nl/en/" target="_blank">Movies that Matter Festival (MtMF)</a> is held annually in The Hague, the Netherlands, and features a wide range of films, including documentaries, fiction films, and short films, all centered on human rights issues such as injustice, women’s rights, and freedom. The festival's aim is to broaden audiences’ perspectives on these issues through cinema. 

The films presented at the festival are selected through the following process: Films are submitted through an open call, and eligible entries must explicitly engage with human rights themes. Final selection decisions are made by the festival’s curatorial team, which holds responsibility for shaping the overall program. As a result, the curatorial process plays an active role in determining which narratives, themes, and regions are made visible. From an analytical perspective, this is significant, as potential curatorial bias may influence patterns of inclusion and exclusion within the festival’s representation of human rights.

## 1.2 Previous research 

Previous research shows that films and film festivals can play an important role in raising awareness about social and political issues. Sheik Mohamed et al. (2024) generally argue that films that address social topics can encourage reflection and discussion among audiences. Furthermore, Bittar (2020) demonstrates how a human rights short film festival in São Paulo contributed to the development of a municipal human rights education policy, suggesting that artistic initiatives can support civic education. Similarly, Priestley (2019) highlights the political role of film festivals, showing how film curation can be used to center Afrolatinx experiences and challenge dominant narratives. Although this highlights the importance of raising awareness of humanitarian issues through film, limited computational research has been conducted on how these themes are represented in film festivals and what the significance might be.

The only available study that conducted computational research on this topic is by Loist (2025). The author notes that most research on film festivals to date has relied on theoretical considerations, prioritising individual qualitative and historical case studies. Therefore, the project was designed to capture data from a broad variety of films screened at a variety of festivals. The project was called *The Circulation Project*, which used empirical big-data approaches from computational cultural analytics and the Digital Humanities to arrive at evidence-based discussions of alternative distribution patterns in festival runs, as well as the positions and connections of festivals within the network. However, as that study concentrates on data from multiple festivals, we differentiate our study by focusing on a single curated film archive and how the film festival distributes human rights themes globally.

## 1.3 Aim of the project 

This project serves both a preservation and a research purpose. From a preservation perspective, we will scrape all 1,202 film pages, including their metadata, from the <a href="https://moviesthatmatter.nl/en/festival/archive/film-archive/#/" target="_blank">Movies that Matter Film Archive</a>, which, to our knowledge, has not been done before. This dataset is rich for preservation purposes, not only because it covers screened films from 2009 to 2026, but the diversity of themes, regions, and years ensures that the Movies that Matter Film Archive captures a comprehensive historical record of human rights–focused cinema, preserving both the films themselves and the festival’s curatorial decisions for future access and study. Although the Movies that Matter Festival Archive also includes pages about the artists and activists who participate in the festival, we do not include this data in our project, as doing so would change our research aim.

From a research perspective, we will use this dataset to conduct computational research based on information from individual film pages. This includes assigned themes, relevant countries, MtMF year, title, film description, director, producer, release year, country of production, duration, spoken languages, subtitle language, production company, and world sales. The goal of this project is to analyze and visualize global human rights issues as represented through the film archive’s thematic classifications. This is significant because it reveals how curated film archives shape the visibility, framing, and geographic distribution of human rights issues, highlighting which themes are emphasized and which may be underrepresented. This analysis could also offer viewers a clear overview of global human rights concerns as represented in the film archive, potentially serving as a resource for the festival’s website.

## 1.4 Research questions

Our main research question is:

-	How are human rights themes distributed and represented across languages, countries, and time in the Movies that Matter Festival Film Archive?
  
Our sub-questions will guide the analysis:

- What are the most frequent languages spoken in our dataset?
- What are the most frequent themes in our dataset?
- What are the most frequent relevant countries in our dataset?
- What are the most frequent countries of production?
- Which themes are most associated with which spoken languages in our dataset?
- How have the themes changed over time (mtmf_year) in our dataset?
- Which top themes are most addressed in top countries? 
- How do the shares of the most common themes in the different regions (Ukraine, United States, United Kingdom, Germany, Mexico, China, Syria) change over time?

## 2.1 Data collection

The dataset for this project was collected from the Movies that Matter Festival Film Archive, covering films screened from 2009 to 2026. However, some films in the archive are missing certain metadata, such as producers, spoken or subtitle languages, or even descriptions. To handle these inconsistencies, the data collection code is designed to fill missing values (NAs) whenever information is absent. This ensures that the resulting dataset does not contain empty lists or columns, which could otherwise produce errors in the analysis part.

All data collection is conducted for educational purposes only. Because the website does not provide a robots.txt file, the code and collected data are stored in a private GitHub repository with access restricted to the professors supervising the project, ensuring responsible and ethical data handling.


## 2.2 Tools 

A range of tools is used to support the collection, cleaning, and structuring of the dataset. Web scraping is performed using **Requests**, **BeautifulSoup**, **Time**, and **Selenium**, while **Pandas** is employed for cleaning, preprocessing, and structuring the data. Each step of the process is documented in a separate **Jupyter Notebook** (Preservation.ipynb) to ensure clarity and reproducibility. The final output for preservation purposes is a structured CSV dataset (df_mtmf), ready for further analysis.

The CSV file (df_mtmf) consists of multiple rows of metadata from the individual film pages, these data will be processed and explored using **Pandas** and **re** (regular expressions), and visualized using **Matplotlib** (matplotlib.pyplot) and **Plotly** (plotly.express and plotly.io). ## 4.1 Limitations

1. Although curatorial bias may play a role in shaping the MtMF Film Archive, we are not able to determine whether this results in the underrepresentation of certain regions, countries, or spoken languages. Assessing such underrepresentation would require a comparative analysis across multiple film archives or festivals, as patterns of absence or dominance can only be meaningfully interpreted in relation to other curated collections. Since this study focuses on a single film archive, our findings should therefore be understood as reflecting the internal thematic and geographic distribution of the Movies that Matter Festival, rather than broader trends in global film production or representation.

2. In the final CSV file (df_movies), the subtitle fields contain languages enclosed in brackets (e.g., ['EN']). During the scraping process, several metadata fields, such as themes, were intentionally stored as lists of strings (e.g., ['justice', 'war']) to account for films associated with multiple themes. However, when the dataframe was exported to a CSV file, these lists were automatically converted into string representations rather than being preserved as list objects. As a result, the bracketed formatting remained in the output. This is why, in the analysis notebook we use regular expressions to get rid of the brackets. Although this formatting could be resolved by adjusting the data structure or rerunning the scraping process, doing so was not feasible within the project timeline, as processing each column required approximately 10 to 40 minutes. Due to these time constraints, the existing outputs were retained.

3. During the scraping process, some films listed the relevant countries as “Universal” or “Africa Continent.” Because these categories do not specify individual countries or regions, they limit the level of geographic detail available for analysis. As this categorization is determined by the festival’s curatorial team, it remains as is.

4. Because relevant countries, countries of production, and spoken languages are not represented in equal proportions, the analysis is not normalized. Normalizing these variables could lead to more balanced results.

## 4.1 Future Research

Future research could focus on analyzing film archives from other film festivals in other parts of the world that address human rights issues (e.g., <a href="https://avhumanrightsfestival.ph/" target="_blank">Active Vista Human Rights Festival</a> in the Philippines), in order to examine whether human rights themes reveal variations in representation across regions. Such comparative approaches would make it possible to identify potential curatorial biases by situating the Movies that Matter Festival within a broader international context.
